```python
import json
import statistics
from collections import Counter
```

```python
def analyze_corpus(name, text_list):
    print(f"--- Analyzing Corpus: {name} ---")
    
    # 1. Average Sequence Length (in characters)
    lengths = [len(text) for text in text_list]
    avg_len = sum(lengths) / len(lengths)
    print(f"Total Documents: {len(text_list)}")
    print(f"Average Sequence Length: {avg_len:.2f} characters")
    
    # 2. Vocabulary Size (Character level for NanoGPT)
    full_text = "".join(text_list)
    vocab = set(full_text)
    print(f"Character Vocabulary Size: {len(vocab)} unique characters")
    
    # 3. Repetition / Noise (Approximated by unique vs total words)
    words = full_text.split()
    unique_words = set(words)
    lexical_diversity = len(unique_words) / len(words) if words else 0
    print(f"Lexical Diversity (Unique/Total Words): {lexical_diversity:.4f} (Lower = More Repetition)\n")
```

```python

# Load Poems (JSONL format)
poems_data = []
with open('/home/naman/Cryptonite-JM-Tasks/GenAI/Task-4/Project/data/poems_pretraining_corpus.jsonl', 'r', encoding='utf-8') as f:
    for line in f:
        data = json.loads(line)
        # Extracting the actual poem content, ignoring the control tokens for the raw metric
        text = data['text'].replace("<|startoftext|>", "").replace("<|endoftext|>", "").strip()
        poems_data.append(text)

# Load Raps (JSON format)
raps_data = []
with open('/home/naman/Cryptonite-JM-Tasks/GenAI/Task-4/Project/data/tagged_rap_dataset_gemma4.json', 'r', encoding='utf-8') as f:
    raps_json = json.load(f)
    for entry in raps_json:
        # Assuming the key containing the text is 'lyrics' based on standard rap datasets
        text = entry.get('lyrics', '')
        if text:
            raps_data.append(text)

# Run Analysis
analyze_corpus("Pre-training Poems (20k)", poems_data)
analyze_corpus("Fine-tuning Raps (2k)", raps_data)
```
